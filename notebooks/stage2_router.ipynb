```python
def classify_query(query):

    query_lower = query.lower()

    # ---------- Equation ----------
    if "=" in query or "formula" in query_lower:

        return "equation"

    # ---------- Diagram ----------
    elif any(word in query_lower for word in [
        "diagram",
        "figure",
        "image",
        "draw"
    ]):

        return "image"

    # ---------- Numerical ----------
    elif any(word in query_lower for word in [
        "calculate",
        "solve",
        "find value",
        "numerical"
    ]):

        return "numerical"

    # ---------- Definition ----------
    elif any(word in query_lower for word in [
        "what is",
        "define",
        "meaning"
    ]):

        return "hybrid"

    # ---------- Why / Explain ----------
    elif any(word in query_lower for word in [
        "why",
        "how",
        "explain",
        "reason"
    ]):

        return "semantic"

    # ---------- Default ----------
    else:

        return "hybrid"
```

```python
queries = [

    "What is inertia?",
    "Why does an object resist motion?",
    "Explain F = ma",
    "Show diagram of force",
    "Solve numerical on acceleration"
]

for q in queries:

    route = classify_query(q)

    print(q, "→", route)
```
