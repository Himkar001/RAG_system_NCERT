```python
import os
from dotenv import load_dotenv
```

```python
load_dotenv()

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

print("Loaded:", GEMINI_API_KEY is not None)
```

```python
def fuse_context(results):

    context_blocks = []

    for r in results:

        text = r.get("text", "")

        context_blocks.append(text)

    return "\n\n".join(context_blocks)
```

```python
def build_prompt(query, retrieved_context):

    prompt = f"""
You are an NCERT Science Tutor.

Answer ONLY using the provided context.

If answer is not found, say:
"I could not find this in the NCERT context."

Context:
{retrieved_context}

Question:
{query}

Provide:
1. Clear explanation
2. Simple language
3. NCERT-grounded answer
"""

    return prompt
```

```python
dummy_results = [

    {
        "text": "Inertia is the tendency of an object to resist change in motion."
    },

    {
        "text": "Newton first law explains inertia."
    }
]
```

```python
context = fuse_context(dummy_results)

print(context)
```

```python
query = "What is inertia?"

prompt = build_prompt(query, context)

print(prompt)
```

```python
import google.generativeai as genai
```

```python
genai.configure(api_key=GEMINI_API_KEY)

model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
response = model.generate_content(prompt)

print(response.text)
```
