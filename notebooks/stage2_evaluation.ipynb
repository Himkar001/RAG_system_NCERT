```python
import pandas as pd
import json
```

```python
evaluation_questions = [

    {
        "question": "What is inertia?",
        "expected_topic": "inertia"
    },

    {
        "question": "Explain Newton first law",
        "expected_topic": "newton"
    },

    {
        "question": "What is momentum?",
        "expected_topic": "momentum"
    },

    {
        "question": "Explain F = ma",
        "expected_topic": "force"
    },

    {
        "question": "Why does an object resist motion?",
        "expected_topic": "inertia"
    }
]
```

```python
def classify_query(query):

    query_lower = query.lower()

    if "=" in query or "formula" in query_lower:
        return "equation"

    elif any(word in query_lower for word in [
        "diagram", "image", "figure"
    ]):
        return "image"

    elif any(word in query_lower for word in [
        "calculate", "solve", "numerical"
    ]):
        return "numerical"

    elif any(word in query_lower for word in [
        "why", "how", "explain"
    ]):
        return "semantic"

    else:
        return "hybrid"
```

```python
results = []

for item in evaluation_questions:

    query = item["question"]

    route = classify_query(query)

    results.append({

        "question": query,
        "route": route,
        "expected_topic": item["expected_topic"]
    })

df = pd.DataFrame(results)

df
```

```python
def hybrid_search(query):

    return [
        {
            "text": f"Retrieved context for: {query}"
        }
    ]
```

```python
evaluation_outputs = []

for item in evaluation_questions:

    query = item["question"]

    retrieval_results = hybrid_search(query)

    top_context = retrieval_results[0]["text"]

    evaluation_outputs.append({

        "question": query,
        "top_chunk": top_context[:300]
    })
```

```python
eval_df = pd.DataFrame(evaluation_outputs)

eval_df.to_csv("../outputs/eval_stage2.csv", index=False)

print("Evaluation saved")
```
