```python
def extract_equation(query):

    query = query.lower()

    patterns = [

        r"[a-z]\s*=\s*[a-z]+",
        r"[a-z]\s*=\s*[a-z]\s*[\+\-\*/]\s*[a-z]+"
    ]

    for pattern in patterns:

        match = re.search(pattern, query)

        if match:

            return match.group()

    return None
```

```python
def equation_lookup(query):

    equation = extract_equation(query)

    if equation is None:

        return "No equation detected"

    equation = equation.replace(" ", "")

    for key in equation_db:

        if key.replace(" ", "") == equation:

            return equation_db[key]

    return "Equation not found"
```

```python
result = equation_lookup("Explain F = ma")

print(result)
```
