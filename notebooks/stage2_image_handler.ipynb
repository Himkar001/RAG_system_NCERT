```python
import fitz
import os
import json
from pathlib import Path
import easyocr
```

```python
reader = easyocr.Reader(['en'])
```

```python
def render_pdf_pages(pdf_path):

    doc = fitz.open(pdf_path)

    image_metadata = []

    for page_index in range(len(doc)):

        page = doc.load_page(page_index)

        pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))

        image_name = f"{Path(pdf_path).stem}_page{page_index+1}.png"

        image_path = os.path.join(IMAGE_OUTPUT, image_name)

        pix.save(image_path)

        # OCR text extraction
        ocr_result = reader.readtext(image_path, detail=0)

        page_text = " ".join(ocr_result)

        image_metadata.append({

            "page": page_index + 1,
            "image_path": image_path,
            "pdf_name": Path(pdf_path).name,
            "ocr_text": page_text.lower()
        })

    return image_metadata
```

```python
pdf_path = "../data/raw/pdfs"
```

```python
pdf_files = list(Path(pdf_path).glob("*.pdf"))

all_images = []

for pdf in pdf_files:

    images = render_pdf_pages(str(pdf))

    all_images.extend(images)

print("Pages Indexed:", len(all_images))
```

```python
with open("../data/processed/image_metadata.json", "w") as f:

    json.dump(all_images, f, indent=4)

print("Metadata Saved")
```

```python
def image_search(query):

    query_lower = query.lower()

    scores = []

    for image in all_images:

        score = 0

        for word in query_lower.split():

            if word in image["ocr_text"]:

                score += 1

        scores.append((score, image))

    scores.sort(reverse=True, key=lambda x: x[0])

    best_match = scores[0]

    if best_match[0] > 0:

        return best_match[1]

    return "No image found"
```

```python
result = image_search("diagram of force")

print(result)
```
